This first section demonstrates basic list indexing and slicing, including positive, negative, and range-based indexing. It also shows how to access elements in nested lists.

```python
#getting all elements prior to end index
x[:3]
```

```python
# This line creates a nested list named 'x'.
# It's a list containing three other lists.
x = [["a", "b", "c"],
     ["d", "e", "f"],
     ["g", "h", "i"]]

# This line prints the entire nested list 'x' to the console.
print(x)

# This line accesses an element within the nested list.
# x[2] accesses the third list (['g', 'h', 'i']).
# [0] then accesses the first element ('g') within that third list.
x[2][0]
```

```python
# This line performs slicing on an element within the nested list.
# x[2] accesses the third list (['g', 'h', 'i']).
# [:2] then slices that list to include elements from the beginning up to (but
# not including) index 2,
# resulting in a new list ['g', 'h'].
x[2][:2]
```

This section covers list manipulation techniques, including modifying elements by index, concatenating lists, adding elements with `append()`, and removing elements with `del()`. It also illustrates the concept of list references and how to create a true copy of a list.

```python
# Create the areas list
areas = ["hallway", 11.25, "kitchen", 18.0, "living room", 20.0, "bedroom"
, 10.75, "bathroom", 9.50]

# Alternative slicing to create downstairs (first 6 elements)
downstairs = areas[:6]
print(downstairs)

# Alternative slicing to create upstairs (last four elements)
upstairs = areas[6:]
print(upstairs)
```

```python
###################################
# List manipulation
###################################
# when you copy a list using "=" you are copying the reference to the list
# thus if y=x, when you change an element in y it will change it in x too.

# If you don't want that to happen and want to do a copy without references,
# then do this y=list(x) OR y=x[:]

# Create the areas list
areas = ["hallway", 11.25, "kitchen", 18.0, "living room", 20.0, "bedroom"
, 10.75, "bathroom", 9.50]

# Correct the bathroom area to 10.50
areas[-1] = 10.50
print(areas)

# Change "living room" to "chill zone"
areas[4] = "chill zone"
print(areas)
```

```python
# Create the areas list and make some changes
areas = ["hallway", 11.25, "kitchen", 18.0, "chill zone", 20.0,
         "bedroom", 10.75, "bathroom", 10.50]

# Add poolhouse data to areas, new list is areas_1 ["poolhouse", 24.5]
areas_1 = areas + ["poolhouse", 24.5]
print(areas_1)
```

```python
# Add garage data to areas_1, new list is areas_2 "garage" and float 15.45
areas_2 = areas_1 + ["garage", 15.45]
print(areas_2)

# delete ["poolhouse", 24.5]
print(areas_2[-4:-2])
del(areas_2[-4:-2])
print(areas_2)
```

```python
# Create list areas
areas = [11.25, 18.0, 20.0, 10.75, 9.50]

# Create areas_copy that reference the same data
areas_copy = areas

# creates a copy that does not reference the same data
areas_copy2 = list(areas) # or areas_copy = areas[:]

# Changes areas_copy, but also areas too!!
areas_copy[0] = 5.0

# Change areas_copy2 value and see that areas does not change
areas_copy2[0] = 1

print(areas)
print(areas_copy)
print(areas_copy2)
```
